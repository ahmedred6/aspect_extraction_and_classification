```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.metrics import accuracy_score, classification_report
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB

import re

RANDOM_STATE = 42
```

```python

df = pd.read_csv(
    "../../ar_reviews_100k.tsv",
    sep="\t",
    names=["label", "text"],
    header=None
)

print(df.head())
print("Loaded samples:", len(df))
```

```python
def normalize_arabic(text):
    if not isinstance(text, str):
        return ""
    text = text.strip()

    # remove tashkeel
    text = re.sub(r"[\u0617-\u061A\u064B-\u0652]", "", text)

    # normalize alef
    text = re.sub(r"[إأآا]", "ا", text)

    # normalize taa marbuta
    text = text.replace("ة", "ه")
    text = re.sub(r'(.)\1+', r'\1', text)

    # normalize ya
    text = re.sub(r"[ىي]", "ي", text)

    return text
df = df[df["label"].isin(["Positive", "Negative", "Mixed"])]

# Remove Mixed class completely
df = df[df["label"] != "Mixed"]

print(df["label"].value_counts())

df["clean_text"] = df["text"].apply(normalize_arabic)

X = df["clean_text"]
y = df["label"]
```

```python

unique_labels = sorted(y.unique())
label_map = {old: new for new, old in enumerate(unique_labels)}
print("Label mapping:", label_map)

y = y.map(label_map)


# =======================================================
# 4. TF-IDF FOR ARABIC
# =======================================================

print("\n[+] Fitting Arabic TF-IDF...")

vectorizer = TfidfVectorizer(
    analyzer="word",
    ngram_range=(1,2),
    max_features=50000
)

X_tfidf = vectorizer.fit_transform(X)

print("TF-IDF shape:", X_tfidf.shape)
```

```python
print(df["label"].value_counts())
```

```python

models = {
    "LogReg": LogisticRegression(max_iter=2000, n_jobs=-1),
    "LinearSVM": LinearSVC(),
    "MultinomialNB": MultinomialNB()
}

kfold = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)
results = {}

print("\n================ MODEL COMPARISON ================\n")

for name, model in models.items():
    print(f"\n===== Evaluating {name} =====")
    fold_acc = []

    for fold_idx, (train_idx, val_idx) in enumerate(kfold.split(X_tfidf, y), start=1):

        Xtr = X_tfidf[train_idx]
        Xval = X_tfidf[val_idx]
        ytr = y.iloc[train_idx]
        yval = y.iloc[val_idx]

        model.fit(Xtr, ytr)
        preds = model.predict(Xval)

        acc = accuracy_score(yval, preds)
        fold_acc.append(acc)
        print(f"  Fold {fold_idx} accuracy: {acc:.4f}")

    results[name] = np.mean(fold_acc)
    print(f"--> {name} Mean CV Accuracy = {results[name]:.4f}")
```

```python
best_model_name = max(results, key=results.get)
print("\n================ BEST MODEL =================")
print(f"Best model: {best_model_name} (Accuracy={results[best_model_name]:.4f})")

if best_model_name == "LogReg":
    base_model = LogisticRegression(max_iter=2000, n_jobs=-1)
    param_grid = {
        "C": [0.1, 0.5, 1.0, 2.0, 5.0],
        "penalty": ["l2"]
    }

elif best_model_name == "LinearSVM":
    base_model = LinearSVC()
    param_grid = {
        "C": [0.1, 0.5, 1.0, 2.0, 5.0]
    }

elif best_model_name == "MultinomialNB":
    base_model = MultinomialNB()
    param_grid = {
        "alpha": [0.1, 0.3, 0.5, 1.0]
    }


print("\n================ GRID SEARCH ================\n")

grid = GridSearchCV(
    estimator=base_model,
    param_grid=param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=-1,
    verbose=1
)

grid.fit(X_tfidf, y)

```

```python
final_model = grid.best_estimator_
print("Best params:", grid.best_params_)


# =======================================================
# 7. FINAL TRAIN/TEST SPLIT (BECAUSE YOUR DATA IS NOT SPLIT)
# =======================================================

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)

final_model.fit(X_train, y_train)
preds = final_model.predict(X_test)

print("\n================ FINAL ARABIC TEST RESULTS ================\n")
print("Test Accuracy:", accuracy_score(y_test, preds))
print("\nClassification Report:\n", classification_report(y_test, preds))
```
